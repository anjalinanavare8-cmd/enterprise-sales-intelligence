```python
df["Net_Sales"] = df["Sales_Amount"] - df["Discount_Amount"]

# Completed orders only
completed_df = df[df["Order_Status"] == "Completed"]

# Summary
print("===== SALES SUMMARY =====")
print("Total Transactions:", len(df))
print("Completed Orders:", len(completed_df))
print("Cancelled Orders:", (df["Order_Status"] == "Cancelled").sum())
print("Total Quantity Sold:", df["Quantity"].sum())
print("Total Gross Sales: ₹", df["Sales_Amount"].sum())
print("Total Discount: ₹", df["Discount_Amount"].sum())
print("Total Net Sales: ₹", df["Net_Sales"].sum())

# Branch sales
print("\n===== SALES BY BRANCH =====")
print(df.groupby("Branch")["Sales_Amount"].sum().sort_values(ascending=False))

# Category sales
print("\n===== SALES BY CATEGORY =====")
print(df.groupby("Category")["Sales_Amount"].sum().sort_values(ascending=False))

# Payment method
print("\n===== SALES BY PAYMENT METHOD =====")
print(df.groupby("Payment_Method")["Sales_Amount"].sum().sort_values(ascending=False))

# Product sales
print("\n===== TOP PRODUCTS =====")
print(df.groupby("Product_Name")["Sales_Amount"].sum().sort_values(ascending=False))

# Data quality
print("\n===== DATA QUALITY =====")
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
print("Negative Quantity:", (df["Quantity"] < 0).sum())
print("Zero Quantity:", (df["Quantity"] == 0).sum())
print("Negative Price:", (df["Unit_Price"] < 0).sum())
print("Zero Price:", (df["Unit_Price"] == 0).sum())
print("Invalid Dates:", df["Transaction_Date"].isna().sum())
```
